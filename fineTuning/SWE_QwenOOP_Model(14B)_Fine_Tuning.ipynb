### Unsloth

```python
from unsloth import FastLanguageModel
import torch

fourbit_models = [
    "unsloth/Qwen3-1.7B-unsloth-bnb-4bit", # Qwen 14B 2x faster
    "unsloth/Qwen3-4B-unsloth-bnb-4bit",
    "unsloth/Qwen3-8B-unsloth-bnb-4bit",
    "unsloth/Qwen3-14B-unsloth-bnb-4bit", # Deberíamos usar este (por esto: https://docs.unsloth.ai/get-started/fine-tuning-guide)
    "unsloth/Qwen3-32B-unsloth-bnb-4bit",

    # 4bit dynamic quants for superior accuracy and low memory use
    "unsloth/gemma-3-12b-it-unsloth-bnb-4bit",
    "unsloth/Phi-4",
    "unsloth/Llama-3.1-8B",
    "unsloth/Llama-3.2-3B",
    "unsloth/orpheus-3b-0.1-ft-unsloth-bnb-4bit" # [NEW] We support TTS models!
] # More models at https://huggingface.co/unsloth

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Qwen3-14B",
    max_seq_length = 2048,   # Context length - can be longer, but uses more memory
    load_in_4bit = True,     # 4bit uses much less memory
    load_in_8bit = False,    # A bit more accurate, uses 2x memory
    full_finetuning = False, # We have full finetuning now!
    # token = "hf_...",      # use one if using gated models
)
```

We now add LoRA adapters so we only need to update 1 to 10% of all parameters!

```python
model = FastLanguageModel.get_peft_model(
    model,
    r = 32,           # Choose any number > 0! Suggested 8, 16, 32, 64, 128
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj",],
    lora_alpha = 32,  # Best to choose alpha = rank or rank*2
    lora_dropout = 0, # Supports any, but = 0 is optimized
    bias = "none",    # Supports any, but = "none" is optimized
    # [NEW] "unsloth" uses 30% less VRAM, fits 2x larger batch sizes!
    use_gradient_checkpointing = "unsloth", # True or "unsloth" for very long context
    random_state = 3407,
    use_rslora = True,   # We support rank stabilized LoRA
    loftq_config = None,  # And LoftQ
)
```

### Data Prep


```python
!pip install -U datasets
from datasets import load_dataset

F2_SB_TR_416 = load_dataset("esquivelgor/F2_SB_TR_416", split = "train")
```

```python
F2_SB_TR_416_SPLIT = F2_SB_TR_416.train_test_split(test_size=0.05, seed=42)

F2_SB_TR_416_TRAIN = F2_SB_TR_416_SPLIT['train']
F2_SB_TR_416_VAL = F2_SB_TR_416_SPLIT['test']
```

```python
F2_SB_TR_416_VAL[0]
```

```python
def generate_conversation(example):
    return {
        "messages": [
            {
                "role": "system",
                "content": (
                    f"SETTING: You are an autonomous programmer working in a command-line interface with a file editor.\n"
                    f"You are currently solving the following issue from a repository.\n\n"
                    f"ISSUE:\n{example['problem_statement']}\n\n"
                    f"You are in the repository's root directory: /{example['repo']}.\n\n"
                    "Your goal is to fix the issue by editing exactly one file — no more. You can inspect files, search through content, and run shell commands or custom tools to assist in your debugging process.\n"
                    "Edit the single file needed to solve the problem and feel free to run any checks or searches to inform your changes.\n\n"
                    "You have access to the following tools. Use **only** these — do not invent new tools or combine multiple commands into one:\n\n"
                    "TOOLS:\n"
                    "- bash: runs normal shell commands like `ls`, `cat`, `cd`\n"
                    "- goto <line_number>: moves the window to a line\n"
                    "- open \"<path>\" [<line_number>]: opens file (optionally at a specific line)\n"
                    "- create <filename>: creates and opens a new file\n"
                    "- scroll_up: scrolls window up 100 lines\n"
                    "- scroll_down: scrolls window down 100 lines\n"
                    "- find_file <file_name> [<dir>]: finds file by name or wildcard pattern\n"
                    "- search_dir <search_term> [<dir>]: searches all files in a directory\n"
                    "- search_file <search_term> [<file>]: searches inside a file (defaults to open file)\n"
                    "- edit <start_line>:<end_line> \\n<replacement_text>\\nend_of_edit: replaces lines in open file (requires exact indentation)\n"
                    "- submit: submits the file\n\n"
                    "IMPORTANT RULES:\n"
                    "- Only one command per response. You must wait for feedback before continuing.\n"
                    "- Do NOT use interactive commands like `vim` or `python`.\n"
                    "- The `edit` command must have proper indentation and must span only the specified lines.\n\n"
                    "Your shell prompt looks like this:\n"
                    "(Open file: <path>) <cwd> $\n\n"
                    "RESPONSE FORMAT:\n"
                    "DISCUSSION\n"
                    "Explain your plan or what you’ll do next.\n"
                    "<command>\n"
                    "some_command_here\n"
                    "</command>\n"
                    "Only one command is allowed per message!"
                )
            },
            {
                "role": "assistant",
                "content": (
                    "DISCUSSION\n"
                    f"I'll start by opening the file `{example['file_path']}` to inspect the relevant code and understand where the issue may be coming from.\n"
                    "<command>\n"
                    f"open \"{example['file_path']}\"\n"
                    "</command>"
                )
            }
        ]
    }

# Apply the mapping
SWEbenchOOPDataset_train = F2_SB_TR_416_TRAIN.map(generate_conversation, remove_columns=F2_SB_TR_416_TRAIN.column_names)
SWEbenchOOPDataset_val = F2_SB_TR_416_VAL.map(generate_conversation, remove_columns=F2_SB_TR_416_VAL.column_names)
```

```python
print(len(SWEbenchOOPDataset_train))
print(len(SWEbenchOOPDataset_val))
```

```python
def format_messages(example):
    messages = example["messages"]
    formatted = ""

    # Merge system and user into a single <|system|> message
    if len(messages) >= 2 and messages[0]["role"] == "system" and messages[1]["role"] == "user":
        system_content = messages[0]["content"]
        user_content = messages[1]["content"]
        merged_system = f"{system_content}\n\n{user_content}"
        formatted += f"<|system|>\n{merged_system}\n"
        assistant_start_index = 2
    else:
        # Fallback: don't merge, just process as-is
        assistant_start_index = 0

    # Process remaining messages (assistant only in this case)
    for message in messages[assistant_start_index:]:
        role = message["role"]
        content = message["content"]
        if role == "assistant":
            formatted += f"<|assistant|>\n{content}\n"

    return {"text": formatted}

# Apply to datasets
SWEbenchOOPDataset_train_form = SWEbenchOOPDataset_train.map(format_messages)
SWEbenchOOPDataset_val_form = SWEbenchOOPDataset_val.map(format_messages)
```

Now let's see how long both datasets are:

```python
print(len(SWEbenchOOPDataset_train_form))
print(len(SWEbenchOOPDataset_val_form))
```

<a name="Train"></a>
### Train the model
Now let's use Huggingface TRL's `SFTTrainer`! More docs here: [TRL SFT docs](https://huggingface.co/docs/trl/sft_trainer). We do 60 steps to speed things up, but you can set `num_train_epochs=1` for a full run, and turn off `max_steps=None`.

```python
from trl import SFTTrainer, SFTConfig
trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = SWEbenchOOPDataset_train_form,
    eval_dataset = SWEbenchOOPDataset_val_form, # Can set up evaluation!
    args = SFTConfig(
        dataset_text_field = "text",
        per_device_train_batch_size = 1,
        gradient_accumulation_steps = 8, # Use GA to mimic batch size!
        warmup_steps = 20,
        num_train_epochs = 5, # Set this for 1 full training run. Default is 2 if commented.
        #max_steps = None,
        learning_rate = 2e-5, # Reduce to 2e-5 for long training runs
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "cosine",
        seed = 3407,
        report_to = "none", # Use this for WandB etc
        load_best_model_at_end=True,
    ),
)
```

```python
# @title Show current memory stats
gpu_stats = torch.cuda.get_device_properties(0)
start_gpu_memory = round(torch.cuda.max_memory_reserved() / 1024 / 1024 / 1024, 3)
max_memory = round(gpu_stats.total_memory / 1024 / 1024 / 1024, 3)
print(f"GPU = {gpu_stats.name}. Max memory = {max_memory} GB.")
print(f"{start_gpu_memory} GB of memory reserved.")
```

Let's train the model! To resume a training run, set `trainer.train(resume_from_checkpoint = True)`

```python
trainer_stats = trainer.train(resume_from_checkpoint = None)
```

```python
from matplotlib import pyplot as plt

train_loss = []
eval_loss = []
steps = []

for log in trainer.state.log_history:
    if 'loss' in log and 'step' in log:
        train_loss.append(log['loss'])
        steps.append(log['step'])
    if 'eval_loss' in log:
        eval_loss.append(log['eval_loss'])

plt.figure(figsize=(10, 6))
plt.plot(steps, train_loss, label='Train Loss', color='blue')
#plt.plot(steps[:len(eval_loss)], eval_loss, label='Eval Loss', color='orange', marker='o')
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Train Loss During Training")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# @title Show final memory and time stats
used_memory = round(torch.cuda.max_memory_reserved() / 1024 / 1024 / 1024, 3)
used_memory_for_lora = round(used_memory - start_gpu_memory, 3)
used_percentage = round(used_memory / max_memory * 100, 3)
lora_percentage = round(used_memory_for_lora / max_memory * 100, 3)
print(f"{trainer_stats.metrics['train_runtime']} seconds used for training.")
print(
    f"{round(trainer_stats.metrics['train_runtime']/60, 2)} minutes used for training."
)
print(f"Peak reserved memory = {used_memory} GB.")
print(f"Peak reserved memory for training = {used_memory_for_lora} GB.")
print(f"Peak reserved memory % of max memory = {used_percentage} %.")
print(f"Peak reserved memory for training % of max memory = {lora_percentage} %.")
```

### Inference
Let's run the model via Unsloth native inference! According to the `Qwen-3` team, the recommended settings for reasoning inference are `temperature = 0.6, top_p = 0.95, top_k = 20`

For normal chat based inference, `temperature = 0.7, top_p = 0.8, top_k = 20`

```python
messages = [
    {"role" : "user", "content" : "Solve (x + 2)^2 = 0."}
]
text = tokenizer.apply_chat_template(
    messages,
    tokenize = False,
    add_generation_prompt = True, # Must add for generation
    enable_thinking = False, # Disable thinking
)

from transformers import TextStreamer
_ = model.generate(
    **tokenizer(text, return_tensors = "pt").to("cuda"),
    max_new_tokens = 256, # Increase for longer outputs!
    temperature = 0.7, top_p = 0.8, top_k = 20, # For non thinking
    streamer = TextStreamer(tokenizer, skip_prompt = True),
)
```

### GGUF / llama.cpp Conversion
To save to `GGUF` / `llama.cpp`, we support it natively now! We clone `llama.cpp` and we default save it to `q8_0`. We allow all methods like `q4_k_m`. Use `save_pretrained_gguf` for local saving and `push_to_hub_gguf` for uploading to HF.

Some supported quant methods (full list on our [Wiki page](https://github.com/unslothai/unsloth/wiki#gguf-quantization-options)):
* `q8_0` - Fast conversion. High resource use, but generally acceptable.
* `q4_k_m` - Recommended. Uses Q6_K for half of the attention.wv and feed_forward.w2 tensors, else Q4_K.
* `q5_k_m` - Recommended. Uses Q6_K for half of the attention.wv and feed_forward.w2 tensors, else Q5_K.

[**NEW**] To finetune and auto export to Ollama, try our [Ollama notebook](https://colab.research.google.com/github/unslothai/notebooks/blob/main/nb/Llama3_(8B)-Ollama.ipynb)
