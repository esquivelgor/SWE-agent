### Import dataset

```python
from datasets import load_dataset, DatasetDict
import os
```

```python
# Define variables
local_path = "./datasets"
path = "SWE-bench/SWE-bench"
split = "train"

# Check if local dataset exists
if os.path.exists(f"{local_path}/{path}_{split}.json"):
    print("Loading dataset from local files...")
    dataset = load_dataset("json", data_files={split: f"{local_path}/{path}_{split}.json"})[split]
else:
    print("Downloading dataset from Hugging Face...")
    dataset = load_dataset(path, split=split)

    # Save to local files (as JSON for simplicity)
    os.makedirs(local_path, exist_ok=True)
    dataset.to_json(f"{local_path}/{path}_{split}.json")
```

----
### Filter patch and instance_id and dump into a json

```python
def filter_data_to_predictions(dataset):
    new_dataset = []

    for row in dataset:
        new_row = {
            "instance_id": row['instance_id'],
            "model": "qwen3:14b",
            "prediction": row['patch']
        }
        new_dataset.append(new_row)

    return new_dataset
```

```python
import json
data = filter_data_to_predictions(dataset)

with open("datasetF1_SBL_T.json", mode="w", encoding="utf-8") as write_file:
    json.dump(data, write_file)
```
